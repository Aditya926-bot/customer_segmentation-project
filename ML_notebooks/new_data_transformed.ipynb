```python
import pandas as pd
import numpy as np
```

```python
new_df = pd.read_csv('/content/drive/MyDrive/ML_Project_2.0/new_trasaction.csv')
print(new_df.shape)
new_df.head()
```

```python
true_dupe_mask = new_df.duplicated(subset=['Invoice ID', 'Line'], keep=False)
true_dupes = new_df[true_dupe_mask]
print("Rows involved in true invoice+line duplication:", true_dupes.shape[0])
true_dupes.sort_values(['Invoice ID','Line']).head(10)
```

```python
before = new_df.shape[0]
new_df= new_df.drop_duplicates(subset=['Invoice ID', 'Line'], keep='first')
after = new_df.shape[0]
print(f"Dropped {before - after} duplicate rows. New shape: {new_df.shape}")
```

```python
duplicates = new_df[new_df.duplicated(keep=False)]
```

```python
print("Rows involved in true duplication:", duplicates.shape[0])
print(duplicates)
```

```python
dups_all = new_df[new_df.duplicated(keep=False)]
dups_all.sort_values(by=list(new_df.columns)).head(20)
```

```python
new_df[new_df['Customer ID'] == 366		]
```

```python
new_df['Date'] = pd.to_datetime(new_df['Date'])
```

```python
new_df.info()
```

```python
sales_df = new_df[new_df['Transaction Type'] == 'Sale'].copy()
sales_df.head()
```

```python
returns_df = new_df[new_df['Transaction Type'] == 'Return'].copy()
returns_df.head()
```

```python
sales_df.shape
```

```python
returns_df.shape
```

```python
analysis_date = new_df['Date'].max() + pd.Timedelta(days=1)
print(analysis_date)
```

```python
last_purchase = sales_df.groupby("Customer ID")["Date"].max().reset_index()
last_purchase.columns = ["Customer ID", "Last_Purchase_Date"]
```

```python
customer_features_df = last_purchase.copy()
customer_features_df["Recency"] = (analysis_date - customer_features_df["Last_Purchase_Date"]).dt.days
customer_features_df.head()
```

```python
frequency = new_df.groupby('Customer ID')['Invoice ID'].nunique().reset_index()
frequency.columns = ['Customer ID', 'Frequency']
```

```python
customer_features_df = customer_features_df.merge(frequency, on='Customer ID', how="left")
customer_features_df.head()
```

```python
first_purchase = sales_df.groupby("Customer ID")["Date"].min().reset_index()
first_purchase.columns = ["Customer ID", "first_Purchase_Date"]
```

```python
customer_features_df = customer_features_df.merge(first_purchase, on='Customer ID', how="left")
customer_features_df.head()
```

```python
unique_products = new_df.groupby('Customer ID')['Product ID'].nunique().reset_index()
unique_products.columns = ['Customer ID', 'Unique_Products']
```

```python
customer_features_df = customer_features_df.merge(unique_products, on='Customer ID', how="left")
customer_features_df.head()
```

```python
net_monetary = new_df.groupby('Customer ID')['Line Total'].sum().reset_index()
net_monetary.columns = ['Customer ID', 'Net_Monetary']
```

```python
customer_features_df = customer_features_df.merge(net_monetary, on='Customer ID', how="left")
customer_features_df.head()
```

```python
customer_features_df['Customer_Lifetime'] = (customer_features_df['Last_Purchase_Date'] - customer_features_df['first_Purchase_Date']).dt.days
customer_features_df.head()
```

```python
customer_features_df['Average_Purchase_Interval'] = customer_features_df['Customer_Lifetime'] / (customer_features_df['Frequency'] - 1).replace(0, np.nan)
customer_features_df.head()
```

```python
gross_monetary = sales_df.groupby('Customer ID')['Line Total'].sum().reset_index()
gross_monetary.columns = ['Customer ID', 'Gross_Monetary']
```

```python
customer_features_df = customer_features_df.merge(gross_monetary, on='Customer ID', how="left")
customer_features_df.head()
```

```python
total_quantity = sales_df.groupby('Customer ID')['Quantity'].sum().reset_index()
total_quantity.columns = ['Customer ID', 'Total_Quantity']
```

```python
customer_features_df = customer_features_df.merge(total_quantity, on='Customer ID', how="left")
customer_features_df.head()
```

```python
avg_unit_price = sales_df.groupby('Customer ID')['Unit Price'].mean().reset_index()
avg_unit_price.columns = ['Customer ID', 'Average_Unit_Price']
```

```python
customer_features_df = customer_features_df.merge(avg_unit_price, on='Customer ID', how="left")
customer_features_df.head()
```

```python
avg_discount_rate = sales_df.groupby('Customer ID')['Discount'].mean().reset_index()
avg_discount_rate.columns = ['Customer ID', 'average_discount_rate']
```

```python
customer_features_df = customer_features_df.merge(avg_discount_rate, on='Customer ID', how="left")
customer_features_df.head()
```

```python
products_with_base_price = new_df[(new_df['Transaction Type'] == 'Sale') & (new_df['Discount'] == 0)] \
    .groupby('Customer ID')['Quantity'].sum().reset_index()
products_with_base_price.columns = ['Customer ID', 'products_with_base_price']
```

```python
customer_features_df = customer_features_df.merge(products_with_base_price, on='Customer ID', how="left")
customer_features_df.head()
```

```python
products_with_discount = new_df[(new_df['Transaction Type'] == 'Sale') & (new_df['Discount'] > 0)] \
    .groupby('Customer ID')['Quantity'].sum().reset_index()
products_with_discount.columns = ['Customer ID', 'products_with_discount']
```

```python
customer_features_df = customer_features_df.merge(products_with_discount, on='Customer ID', how="left")
customer_features_df.head()
```

```python
return_count = returns_df.groupby('Customer ID')['Invoice ID'].nunique().reset_index()
return_count.columns = ['Customer ID', 'Return_Count']
```

```python
customer_features_df = customer_features_df.merge(return_count, on='Customer ID', how="left")
customer_features_df.head()
```

```python
returned_quantity = returns_df.groupby('Customer ID')['Quantity'].sum().reset_index()
returned_quantity.columns = ['Customer ID', 'Returned_Quantity']
```

```python
customer_features_df = customer_features_df.merge(returned_quantity, on='Customer ID', how="left")
customer_features_df.head()
```

```python
return_value = returns_df.groupby('Customer ID')['Line Total'].sum().abs().reset_index()
return_value.columns = ['Customer ID', 'Return_Value']
```

```python
customer_features_df = customer_features_df.merge(return_value, on='Customer ID', how="left")
customer_features_df.head()
```

```python
sales_df['DayOfWeek'] = sales_df['Date'].dt.day_name()
sales_df['MonthName'] = sales_df['Date'].dt.month_name()
sales_df.head()
```

```python
fav_day = sales_df.groupby('Customer ID')['DayOfWeek'].agg(lambda x: x.mode()[0]).reset_index()
fav_day.columns = ['Customer ID', 'Favorite_Shopping_Day_in_week']
```

```python
customer_features_df = customer_features_df.merge(fav_day, on='Customer ID', how="left")
customer_features_df.head()
```

```python
fav_month = sales_df.groupby('Customer ID')['MonthName'].agg(lambda x: x.mode()[0]).reset_index()
fav_month.columns = ['Customer ID', 'Favorite_Shopping_month_in_year']
```

```python
customer_features_df = customer_features_df.merge(fav_month, on='Customer ID', how="left")
customer_features_df.head()
```

```python
sales_df['YearMonth'] = sales_df['Date'].dt.to_period('M')
sales_df.head()
```

```python
monthly = sales_df.groupby(['Customer ID', 'YearMonth'])['Line Total'].sum().reset_index()

monthly_mean = monthly.groupby('Customer ID')['Line Total'].mean().reset_index()
monthly_mean.columns = ['Customer ID', 'Monthly_Spending_Mean']

monthly_mean.head()
```

```python
customer_features_df = customer_features_df.merge(monthly_mean, on='Customer ID', how="left")
customer_features_df.head()
```

```python
monthly = sales_df.groupby(['Customer ID', 'YearMonth'])['Line Total'].sum().reset_index()

monthly_std = monthly.groupby('Customer ID')['Line Total'].std().reset_index()
monthly_std.columns = ['Customer ID', 'Monthly_Spending_Std']

monthly_std.head()
```

```python
customer_features_df = customer_features_df.merge(monthly_std, on='Customer ID', how="left")
customer_features_df.head()
```

```python
monthly = sales_df.groupby(['Customer ID', 'YearMonth'])['Line Total'].sum().reset_index()
monthly = monthly.sort_values(['Customer ID', 'YearMonth'])

def calc_trend(g):
    if len(g) < 2:
        return 0.0
    x = np.arange(len(g))
    y = g['Line Total'].values
    return np.polyfit(x, y, 1)[0]

spending_trend = monthly.groupby('Customer ID').apply(calc_trend).reset_index()
spending_trend.columns = ['Customer ID', 'Spending_Trend']

spending_trend.head()
```

```python
customer_features_df = customer_features_df.merge(spending_trend, on='Customer ID', how="left")
customer_features_df.head()
```

```python
customer_features_df.info()
```

```python
features = [
    "products_with_base_price",
    "products_with_discount",
    "Return_Count",
    "Returned_Quantity",
    "Return_Value"
]

customer_features_df[features] = customer_features_df[features].fillna(0)
```

```python
customer_features_df.info()
```

```python
customer_features_df.head(10)
```

```python
customer_features_df['Is_One_Time_Customer'] = (customer_features_df['Frequency'] == 1).astype(int)
customer_features_df['Average_Purchase_Interval'] = customer_features_df['Average_Purchase_Interval'].fillna(0)
customer_features_df['Monthly_Spending_Std'] = customer_features_df['Monthly_Spending_Std'].fillna(0)
```

```python
customer_features_df.head(10)
```

```python
sale_invoice_count = sales_df.groupby('Customer ID')['Invoice ID'].nunique().reset_index()
sale_invoice_count.columns = ['Customer ID', 'Sale_Frequency']

customer_features_df = customer_features_df.merge(sale_invoice_count, on='Customer ID', how='left')
customer_features_df['Sale_Frequency'] = customer_features_df['Sale_Frequency'].fillna(0)

customer_features_df['average_order_value'] = customer_features_df['Gross_Monetary'] / customer_features_df['Sale_Frequency']
customer_features_df['Return_Rate'] = customer_features_df['Return_Count'] / customer_features_df['Sale_Frequency']

customer_features_df.head()
```
